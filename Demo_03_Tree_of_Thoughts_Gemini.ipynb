```python
# -*- coding: utf-8 -*-
"""
Demo_03_Tree_of_Thoughts_Gemini.py


=====================================================
Tree-of-Thought Prompting using Gemini
=====================================================

Objective:
To perform tree-of-thought prompting using Google's
Gemini model, enabling exploration of multiple
reasoning branches and convergence on the best
solution.

=====================================================
"""

import os
import google.generativeai as genai



```

```python
# =====================================================
# Step 1: Set up the Gemini API key
# =====================================================

# Replace with your actual Gemini API key
os.environ["GOOGLE_API_KEY"] = ""

# Configure Gemini
genai.configure(api_key=os.getenv("GOOGLE_API_KEY"))


```

```python
# =====================================================
# Step 2: Define a function to get completion
# =====================================================

"""
This helper function sends a prompt to the Gemini model
and retrieves a deterministic response suitable for
tree-of-thought reasoning.
"""

def get_completion(prompt, model="gemini-2.5-flash"):
    try:
        gemini_model = genai.GenerativeModel(model)

        response = gemini_model.generate_content(
            prompt,
            generation_config={
                "temperature": 0,
            }
        )

        return response.text

    except Exception as e:
        return f"Error: {str(e)}"

```

```python

# =====================================================
# Step 3: Define Tree-of-Thought Prompts
# =====================================================

# -----------------------------------------------------
# Example 1: Maximizing area with given fencing
# -----------------------------------------------------

prompt = """
Solve the problem:
A farmer has 100 meters of fencing and wants to enclose
the maximum area for his rectangular field.

Let's explore multiple reasoning paths:

1. Square configuration:
- Each side = 100 / 4 = 25 meters
- Area = 25 × 25 = 625 square meters

2. Rectangular configuration (4:1 ratio):
- Length = 40 meters, Width = 10 meters
- Area = 400 square meters

3. Explore other possible ratios and compare areas.

Evaluate all paths and conclude the best dimensions
for maximum area.
"""

print("=== Tree-of-Thought Example 1 ===\n")
response = get_completion(prompt)
print(response)

```

```python

# -----------------------------------------------------
# Example 2: Legal case analysis with reasoning branches
# -----------------------------------------------------

prompt = """
Analyze the legal case by exploring multiple precedents
and reasoning branches.

1. Precedent 1: Plaintiff won
- Branch A: Defendant found negligent
    - Sub-branch A1: Clear evidence of negligence
    - Sub-branch A2: Breach of duty established

2. Precedent 2: Defendant won
- Branch B: No negligence found
    - Sub-branch B1: Insufficient evidence
    - Sub-branch B2: Plaintiff assumed the risk

3. Precedent 3: Mixed outcome
- Branch C: Shared fault
    - Sub-branch C1: Reduced damages
    - Sub-branch C2: Split liability

4. Based on the current case facts, identify the
most likely outcome and justify it.
"""

print("\n---------------------------------\n")
print("=== Tree-of-Thought Example 2 ===\n")
response = get_completion(prompt)
print(response)


```

```python
# -----------------------------------------------------
# Example 3: Analytical reasoning problem (12 balls)
# -----------------------------------------------------

prompt = """
You have 12 identical-looking balls, but one is either
heavier or lighter. You have a balance scale and can
use it only three times.

At each step:
- Explore multiple possible actions
- Evaluate promising and unhelpful paths
- Prune ineffective branches

After exploring the reasoning tree, provide the best
solution and explain why it works.
"""

print("\n---------------------------------\n")
print("=== Tree-of-Thought Example 3 ===\n")
response = get_completion(prompt)
print(response)

```

```python

# -----------------------------------------------------
# Example 4: Planning using a reasoning tree
# -----------------------------------------------------

prompt = """
Let's plan a weekend trip by exploring multiple options.

Option 1: Go to the mountains
- Branch A: Good weather
    - Hiking
    - Visit a lake
- Branch B: Bad weather
    - Relax in a cabin
    - Visit museums

Option 2: Go to the beach
- Branch C: Sunny weather
    - Swimming
    - Beach activities
- Branch D: Rainy weather
    - Aquarium visits
    - Shopping

Considering weather conditions in Goa during August,
evaluate all branches and decide the best option.
"""

print("\n---------------------------------\n")
print("=== Tree-of-Thought Example 4 ===\n")
response = get_completion(prompt)
print(response)


```
