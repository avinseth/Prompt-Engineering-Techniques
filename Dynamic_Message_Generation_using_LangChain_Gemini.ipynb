```python
# -*- coding: utf-8 -*-
"""
Demo_07_Dynamic_Message_Generation_Gemini.py


=====================================================
Dynamic Message Generation using LangChain + Gemini
=====================================================

Objective:
To dynamically generate and manage conversation messages
using LangChain message templates and placeholders,
enabling context-aware conversational AI workflows
with Gemini.

=====================================================
"""

import os
import google.generativeai as genai

from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.messages import HumanMessage, AIMessage


# =====================================================
# Step 1: Set up Gemini API key
# =====================================================

# Replace with your Gemini API key
os.environ["GOOGLE_API_KEY"] = "AIzaSyBIhBjovwr8w8VapSpiUcJn-qRB_D183zo"

genai.configure(api_key=os.getenv("GOOGLE_API_KEY"))


```

```python

# =====================================================
# Step 2: Gemini completion function
# =====================================================

def get_completion(prompt: str, model="gemini-2.5-flash") -> str:
    """
    Sends a formatted prompt to Gemini and returns response.
    """
    try:
        gemini_model = genai.GenerativeModel(model)

        response = gemini_model.generate_content(
            prompt,
            generation_config={
                "temperature": 0.7,
            }
        )

        return response.text

    except Exception as e:
        return f"Error: {str(e)}"


```

```python
# =====================================================
# Step 3: Define message templates
# =====================================================

simple_prompt = "The {subject} is strong in this one."
human_prompt = "Summarize our conversation so far in {word_count} words."


simple_message_template = ChatPromptTemplate.from_template(simple_prompt)
human_message_template = ChatPromptTemplate.from_template(human_prompt)


```

```python
# =====================================================
# Step 4: Create chat prompt with placeholders
# =====================================================

chat_prompt = ChatPromptTemplate.from_messages([
    MessagesPlaceholder(variable_name="conversation"),
    simple_message_template,
    human_message_template
])


```

```python
# =====================================================
# Step 5: Define conversation messages
# =====================================================

human_message = HumanMessage(
    content="What's the best way to learn a new language?"
)

ai_message = AIMessage(
    content="""\
1. Immerse yourself in the language.
2. Practice consistently.
3. Use language learning apps.
"""
)


```

```python
# =====================================================
# Step 6: Generate conversation messages
# =====================================================

conversation = chat_prompt.format_prompt(
    conversation=[human_message, ai_message],
    subject="Force",
    word_count="10"
).to_messages()


print("---- Conversation 1 ----")
for message in conversation:
    print(message.content)


```

```python
# =====================================================
# Step 7: Modify prompts dynamically
# =====================================================

simple_prompt = "The {subject} is fascinating to study."
human_prompt = "Summarize our conversation so far in {word_count} words."

simple_message_template = ChatPromptTemplate.from_template(simple_prompt)
human_message_template = ChatPromptTemplate.from_template(human_prompt)

chat_prompt = ChatPromptTemplate.from_messages([
    MessagesPlaceholder(variable_name="conversation"),
    simple_message_template,
    human_message_template
])

human_message = HumanMessage(
    content="What happens inside a black hole?"
)

ai_message = AIMessage(
    content="Inside a black hole, gravity becomes infinitely strong."
)

conversation = chat_prompt.format_prompt(
    conversation=[human_message, ai_message],
    subject="Black Hole",
    word_count="10"
).to_messages()

print("\n---- Conversation 2 ----")
for message in conversation:
    print(message.content)

```

```python

# =====================================================
# Step 8: Prompt without conversation placeholder
# =====================================================

simple_prompt = "The {subject} is fascinating to study."
human_prompt = "Summarize our discussion in {word_count} words."

simple_message_template = ChatPromptTemplate.from_template(simple_prompt)
human_message_template = ChatPromptTemplate.from_template(human_prompt)

chat_prompt = ChatPromptTemplate.from_messages([
    simple_message_template,
    human_message_template
])

formatted_prompt = chat_prompt.format_prompt(
    subject="Black Holes",
    word_count="30"
)

prompt_string = formatted_prompt.to_string()

print("\n---- Formatted Prompt ----")
print(prompt_string)

```

```python

# =====================================================
# Step 9: Send formatted prompt to Gemini
# =====================================================

response = get_completion(prompt_string)

print("\n---- Gemini AI Response ----")
print(response)


```
