```python
#      -*- coding: utf-8 -*-
"""
Demo_01_Chain_of_Thoughts_Gemini.py

Gemini (Google Generative AI)

=====================================================
Chain-of-Thought Prompting using Gemini
=====================================================

Objective:
To perform chain-of-thought prompting using Google's
Gemini model, enabling step-by-step reasoning for
logical, analytical, and problem-solving tasks.

=====================================================
"""

import os
import google.generativeai as genai

```

```python

# =====================================================
# Step 2: Define a function to get completion
# =====================================================

"""
This function sends a prompt to the Gemini model
and returns a deterministic response suitable for
reasoning and chain-of-thought tasks.
"""

def get_completion(prompt, model="gemini-2.5-flash-lite"):
    try:
        gemini_model = genai.GenerativeModel(model)

        response = gemini_model.generate_content(
            prompt,
            generation_config={
                "temperature": 0,
            }
        )

        return response.text

    except Exception as e:
        return f"Error: {str(e)}"

```

```python

# =====================================================
# Step 3: Chain-of-Thought Prompt Sequence
# =====================================================

prompts = [
    "Imagine you are a detective trying to solve a mystery.",
    "You arrive at the crime scene and start looking for clues.",
    "You find a strange object at the crime scene. What is it?",
    "How does this object relate to the crime?",
    "Who do you think is the suspect and why?"
]

print("=== Detective Chain-of-Thought Example ===\n")

for prompt in prompts:
    response = get_completion(prompt)
    print(f"Prompt: {prompt}")
    print(f"Response: {response}")
    print()


```

```python
# =====================================================
# Step 4: Analyze the Output
# =====================================================

# Example 1: Sentiment analysis with reasoning steps

prompt = """
Let's analyze the sentiment of the review step by step.

1. Identify the positive aspect of the review and give a score out of 10.
2. Identify the negative aspect of the review and give a score out of 10.
3. Weigh the positive and negative aspects to determine the overall sentiment.
4. Provide the final sentiment classification with justification.

Review: "The product is very well designed product"
"""

print("=== Sentiment Analysis with Reasoning ===\n")
response = get_completion(prompt)
print(response)
```

```python


# =====================================================
# Example 2: Logical sequence reasoning
# =====================================================

prompt = """
Let's sort the values of the list step by step.

1. Start with the unsorted list.
2. Compare each element and find the smallest value.
3. Place the smallest value in the first position.
4. Repeat the process for all remaining elements.
5. Provide the final sorted list.

Sort the list: [3, 1, 4, 6, 5, 9, 2]
"""

print("\n=== Logical Sorting Reasoning ===\n")
response = get_completion(prompt)
print(response)

```

```python

# =====================================================
# Example 3: Analytical problem-solving reasoning
# =====================================================

prompt = """
You have 12 identical-looking balls, but one is either heavier or lighter.
You have a balance scale and can only use it three times.

Explain step-by-step how you can find the odd ball and
determine whether it is heavier or lighter.
Think carefully and explain your reasoning before
giving the final answer.
"""

print("\n=== Analytical Problem-Solving ===\n")
response = get_completion(prompt)
print(response)


```
