### __Step 1: Tokenization__


- **What is Tokenization?**  
  Tokenization is the process of breaking text into smaller units called *tokens*, such as words or subwords.  
  This helps in text analysis, NLP preprocessing, and model training.  


```python
text_data = [
    "AI is transforming the world.",
    "Natural language processing is a subset of AI.",
    "Deep learning and machine learning are popular AI techniques.",
    "AI applications are diverse and growing rapidly."
]
```

```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')


# Tokenize the text data
from nltk.tokenize import word_tokenize

tokenized_data = [word_tokenize(sentence) for sentence in text_data]
```

```python
tokenized_data
```

- Each sentence is split into individual words, preparing the data for further processing.

## __Step 2: Bag of Words (BoW) Vectorization__

- What is Bag of Words? \
    Bag of Words (BoW) converts text into numerical form by counting the occurrence of each word across the dataset.
    It ignores grammar and word order, focusing purely on word frequency.

```python
from sklearn.feature_extraction.text import CountVectorizer

# Initialize the CountVectorizer
vectorizer = CountVectorizer()

# Fit and transform the text data
vectorized_data = vectorizer.fit_transform(text_data)

# Get the feature names (tokens)
feature_names = vectorizer.get_feature_names_out()

# Convert to array for better visualization
vector_array = vectorized_data.toarray()

feature_names, vector_array
```

- Each row in the matrix represents a sentence, and each column represents a unique word in the corpus.

## __Step 3: One-Hot Encoding__

- What is One-Hot Encoding? \
    One-Hot Encoding represents each unique word as a binary vector.
    Each word gets a unique index, and a value of 1 indicates the presence of that word in a given sentence.

```python
from sklearn.preprocessing import OneHotEncoder
import numpy as np

# Sample Text Dataset
text_data = [
    "AI is transforming the world.",
    "Natural language processing is a subset of AI.",
    "Deep learning and machine learning are popular AI techniques.",
    "AI applications are diverse and growing rapidly."
]

# Tokenize
tokenized_data = [word_tokenize(sentence.lower()) for sentence in text_data]

# Flatten tokens
flattened_tokens = [word for sentence in tokenized_data for word in sentence]

# Vocabulary
vocabulary = sorted(list(set(flattened_tokens)))

# OneHotEncoder
encoder = OneHotEncoder(sparse_output=False)
encoder.fit(np.array(vocabulary).reshape(-1, 1))

# TRUE one-hot encoding (per word)
true_one_hot = {
    word: encoder.transform([[word]])[0]
    for word in vocabulary
}

# Example: one-hot for "ai"
true_one_hot['natural']
```

## __Step 4: Word Embeddings (Word2Vec)__

- What is Word2Vec? \
    Word2Vec converts words into dense vector representations where words with similar meanings are positioned closer in the vector space.
    This helps models understand contextual relationships between words.

```python
!pip install gensim
```

```python
import nltk
from gensim.models import Word2Vec

nltk.download('punkt')

# Sample Text Dataset
text_data = [
    "AI is transforming the world.",
    "Natural language processing is a subset of AI.",
    "Deep learning and machine learning are popular AI techniques.",
    "AI applications are diverse and growing rapidly."
]

# Tokenize the text data
tokenized_data = [word_tokenize(sentence.lower()) for sentence in text_data]

# Train the Word2Vec model
model = Word2Vec(sentences=tokenized_data, vector_size=100, window=5, min_count=1, workers=4)

# Get the vocabulary
vocabulary = list(model.wv.index_to_key)

# Get word embeddings for the vocabulary
word_embeddings = {word: model.wv[word] for word in vocabulary}

# Example: Get embedding for the word 'ai'
embedding_ai = word_embeddings['ai']

vocabulary, embedding_ai
```

- Word embeddings capture semantic meaning — for example, words like AI, machine, and learning will have similar vector representations.
