```python
# -*- coding: utf-8 -*-
"""
Demo_02_Self_Consistency_Gemini.py

Gemini (Google Generative AI)

=====================================================
Self-Consistency Prompting using Gemini
=====================================================

Objective:
To perform self-consistency prompting using Google's
Gemini model by generating multiple reasoning paths
and selecting the most consistent and logical answer.

=====================================================
"""

import os
import google.generativeai as genai

```

```python

# =====================================================
# Step 1: Set up the Gemini API key
# =====================================================

# Replace with your actual Gemini API key
os.environ["GOOGLE_API_KEY"] = ""

# Configure Gemini
genai.configure(api_key=os.getenv("GOOGLE_API_KEY"))



```

```python
# =====================================================
# Step 2: Define a function to get completion
# =====================================================

"""
This function sends a prompt to the Gemini model
and retrieves a deterministic response suitable
for self-consistency reasoning.
"""

def get_completion(prompt, model="gemini-2.5-flash"):
    try:
        gemini_model = genai.GenerativeModel(model)

        response = gemini_model.generate_content(
            prompt,
            generation_config={
                "temperature": 0,
            }
        )

        return response.text

    except Exception as e:
        return f"Error: {str(e)}"

```

```python

# =====================================================
# Step 3: Evaluate the Output (Self-Consistency)
# =====================================================

# Example 1: Weight comparison using multiple reasoning paths

prompt = """
Let's consider which is heavier: 1000 feathers or a 30-pound weight.
I'll think through this in a few different ways and then decide which
answer seems most consistent.

1. First line of reasoning:
A single feather is very light. Even 1000 feathers may still be lighter
than a 30-pound weight.

2. Second line of reasoning:
1000 is a large number, and adding many feathers together might make
them heavy.

3. Third line of reasoning:
The average weight of a feather is extremely small, so even 1000 feathers
would not add up to 30 pounds.

Considering these reasonings, the most consistent answer is:
"""

print("=== Self-Consistency Example 1 ===\n")
response = get_completion(prompt)
print(response)

```

```python

# =====================================================
# Example 2: Word problem with multiple interpretations
# =====================================================

prompt = """
A farmer has 17 sheep. All but 9 run away. How many are left?

1. "All but 9 ran away" means 9 sheep are left.

2. "All but 9" means only 9 stayed behind.

3. Subtracting 17 - 9 gives 8, but this interpretation is incorrect.

Considering these reasonings, the most consistent answer is:
"""

print("\n---------------------------------\n")
print("=== Self-Consistency Example 2 ===\n")
response = get_completion(prompt)
print(response)

```

```python

# =====================================================
# Example 3: Math logic problem verification
# =====================================================

prompt = """
I will solve the following problem in several different ways
and check which answer appears most consistently.

Problem:
There were 15 apples, and you took away 4.
How many apples do you have?

1. First approach:
15 - 4 = 11 apples (incorrect interpretation)

2. Second approach:
If I take away 4 apples, then I have 4 apples (correct)

3. Third approach:
Taking away 4 apples means I possess 4 apples (correct)

4. Fourth approach:
Subtracting gives 11, but that answers a different question

Based on these reasoning paths, the most consistent answer is:
"""

print("\n---------------------------------\n")
print("=== Self-Consistency Example 3 ===\n")
response = get_completion(prompt)
print(response)

```
